```python
from litellm import completion
from typing import List, Dict


def generate_response(messages: List[Dict]) -> str:
    """Call LLM to get response"""
    response = completion(
        model="gemini/gemini-3-flash-preview",
        messages=messages,
        max_tokens=1024
    )
    return response.choices[0].message.content


function_request = input("What Python function would you like to create? ")

messages = [
    {"role": "system", "content": "You are an expert Python developer. Write clear, beginner-friendly Python functions."},
    {"role": "user", "content": function_request}
]

response = generate_response(messages)
print(response)
```
